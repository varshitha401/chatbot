```
import pandas as pd
import spacy

nlp = spacy.load("en_core_web_sm")
```

```
import pandas as pd
import zipfile
import os

# Define the path to the zip file and the name of the CSV file within it
zip_file_path = "/content/archive (1).zip"
csv_file_in_zip = "Resume/Resume.csv"

# Check if the zip file exists
if not os.path.exists(zip_file_path):
    raise FileNotFoundError(f"The file '{zip_file_path}' does not exist. Please ensure the 'archive.zip' file is uploaded to '/content/'.")

# Extract the specific CSV file from the zip archive
try:
    with zipfile.ZipFile(zip_file_path, 'r') as zf:
        # Check if the CSV file exists within the zip archive
        if csv_file_in_zip not in zf.namelist():
            raise FileNotFoundError(f"The specified CSV file '{csv_file_in_zip}' was not found inside '{zip_file_path}'. Available files are: {zf.namelist()}.")
        with zf.open(csv_file_in_zip) as csv_file:
            df = pd.read_csv(csv_file)
except zipfile.BadZipFile:
    raise ValueError(f"The file '{zip_file_path}' is not a valid ZIP file or is corrupted. Please ensure you have uploaded a correct and uncorrupted ZIP archive.")
except Exception as e:
    # Catch any other unexpected errors during processing
    raise e
```

```
def extract_entitles(text):

  doc = nlp(text)

  names =[]
  orgs = []

  for ent in doc.ents:
    if ent.label_ == "PERSON":
      names.append(ent.text)
    elif ent.label_=="ORG":
      orgs.append(ent.text)

  return names, orgs
```

```
for text in df['Resume_str'].head(5):

  names,orgs = extract_entitles(text)

  print("names:", names)
  print("Organizations:", orgs)
  print("------------")
```
